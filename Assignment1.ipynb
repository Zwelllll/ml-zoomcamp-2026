```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)

url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)
df.head()
```

```python
# Q2 — number of records
print("Q2 records:", len(df))

# Q3 — number of fuel types
print("Q3 fuel types:", df["fuel_type"].nunique())

# Q4 — columns containing missing values
print("Q4 columns with missing values:", df.isna().any().sum())

# Q5 — maximum fuel efficiency for cars from Asia
asia = df[df["origin"] == "Asia"]
print("Q5 max MPG (Asia):", asia["fuel_efficiency_mpg"].max())

# Q6 — fill missing horsepower with its most frequent value
median_before = df["horsepower"].median()
most_frequent = df["horsepower"].mode()[0]
horsepower_filled = df["horsepower"].fillna(most_frequent)
median_after = horsepower_filled.median()

print("Q6 median before:", median_before)
print("Q6 most frequent:", most_frequent)
print("Q6 median after:", median_after)

# Q7 — matrix calculation
X = asia[["vehicle_weight", "model_year"]].head(7).to_numpy()
XTX = X.T @ X
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = np.linalg.inv(XTX) @ X.T @ y

print("Q7 sum of weights:", w.sum())
```
